```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

def generate_conversion_sankey(df, selected_A, selected_B, gender='All', weight_class='All', 
                               smart_weight_filter=True, grouping='default', b_target='default',
                               european_only=False, dutch_only=False,
                               a_start_year=None, a_end_year=None, 
                               b_start_year=None, b_end_year=None,
                               min_birth_year=None,
                               print_conversion_table=True, plot_type='both',
                               export_html_path=None):
    """
    Genereert Sankey diagrammen en slaat deze optioneel op als HTML.
    Volledig vertaald naar het Nederlands en NOC*NSF huisstijl.
    Visuele update: Bugfix in kleurtoewijzing zorgt voor perfect gescheiden kleuren in de flows.
    """
    
    # ------------------------------------------------------------------
    # 1. PAS BASIS FILTERS TOE
    # ------------------------------------------------------------------
    filtered_df = df.copy()
    filtered_df = filtered_df.dropna(subset=['Person'])
    filtered_df['Year_num'] = pd.to_numeric(filtered_df['Year'], errors='coerce')
    title_filters = []

    if min_birth_year is not None:
        birth_years = pd.Series(np.nan, index=filtered_df.index)
        
        if 'DoB' in filtered_df.columns:
            birth_years = pd.to_datetime(filtered_df['DoB'], errors='coerce').dt.year
            
        if 'YoB' in filtered_df.columns:
            yob_numeric = pd.to_numeric(filtered_df['YoB'], errors='coerce')
            birth_years = birth_years.fillna(yob_numeric)
            
        if 'Age' in filtered_df.columns:
            age_numeric = pd.to_numeric(filtered_df['Age'], errors='coerce')
            calc_year = filtered_df['Year_num'] - age_numeric
            birth_years = birth_years.fillna(calc_year)
            
        filtered_df['Birth_Year_num'] = birth_years
        
        invalid_mask = filtered_df['Birth_Year_num'] < min_birth_year
        invalid_persons = filtered_df[invalid_mask]['Person'].unique()
        filtered_df = filtered_df[~filtered_df['Person'].isin(invalid_persons)]
        title_filters.append(f"Geboren ≥ {min_birth_year}")
    
    if gender and gender.lower() != 'all':
        filtered_df = filtered_df[filtered_df['Gender'].astype(str).str.lower() == gender.lower()]
        nl_gender = "Mannen" if gender.lower() in ['m', 'men', 'male'] else ("Vrouwen" if gender.lower() in ['f', 'w', 'women', 'female'] else gender)
        title_filters.append(f"Geslacht: {nl_gender}")

    if dutch_only:
        dutch_mask = (
            filtered_df['Country'].astype(str).str.lower().isin(['netherlands', 'the netherlands', 'ned']) |
            (filtered_df['Country Code'].astype(str).str.upper() == 'NED') |
            filtered_df['Nationality'].astype(str).str.upper().isin(['NED', 'NETHERLANDS', 'DUTCH'])
        )
        filtered_df = filtered_df[filtered_df['Person'].isin(filtered_df[dutch_mask]['Person'].unique())]
        title_filters.append("Nederlandse Atleten")
    elif european_only:
        eju_codes = ['ALB', 'AND', 'ARM', 'AUT', 'AZE', 'BLR', 'BEL', 'BIH', 'BUL', 'CRO', 'CYP', 'CZE', 
                     'DEN', 'EST', 'FIN', 'FRA', 'GEO', 'GER', 'GBR', 'GRE', 'HUN', 'ISL', 'IRL', 'ISR', 
                     'ITA', 'KOS', 'LAT', 'LIE', 'LTU', 'LUX', 'MLT', 'MDA', 'MON', 'MNE', 'NED', 
                     'MKD', 'NOR', 'POL', 'POR', 'ROU', 'RUS', 'SMR', 'SRB', 'SVK', 'SLO', 'ESP', 
                     'SWE', 'SUI', 'TUR', 'UKR', 'AIN']
        euro_mask = (filtered_df['Continent'].astype(str).str.lower() == 'europe') | \
                    (filtered_df['Country Code'].astype(str).str.upper().isin(eju_codes)) | \
                    (filtered_df['Nationality'].astype(str).str.upper().isin(eju_codes))
        filtered_df = filtered_df[filtered_df['Person'].isin(filtered_df[euro_mask]['Person'].unique())]
        title_filters.append("Europese Atleten")

    # ------------------------------------------------------------------
    # 2. PAS EXPLICIETE TIJDSLIJN FILTERS TOE
    # ------------------------------------------------------------------
    if a_end_year is not None:
        mask_future_a = (filtered_df['Comp.SetDetail'].isin(selected_A)) & (filtered_df['Year_num'] > a_end_year)
        filtered_df = filtered_df[~mask_future_a]
        title_filters.append(f"Toernooi 1 eindigt {a_end_year}")

    if b_end_year is not None:
        mask_future_b = (filtered_df['Comp.SetDetail'].isin(selected_B)) & (filtered_df['Year_num'] > b_end_year)
        filtered_df = filtered_df[~mask_future_b]
        title_filters.append(f"Toernooi 2 eindigt {b_end_year}")

    if filtered_df.empty:
        print("Geen data beschikbaar na het toepassen van de filters.")
        return

    # ------------------------------------------------------------------
    # 3. DATA EXTRACTIE HELPER
    # ------------------------------------------------------------------
    def map_rank_to_placement(rank):
        rank_str = str(rank).strip().lower()
        if rank_str in ['1', '1.0', '1st']: return '1e'
        if rank_str in ['2', '2.0', '2nd']: return '2e'
        if rank_str in ['3', '3.0', '3rd']: return '3e'
        if rank_str in ['5', '5.0', '5th']: return '5e'
        if rank_str in ['7', '7.0', '7th']: return '7e'
        return 'Deelname'

    df_A_raw = filtered_df[filtered_df['Comp.SetDetail'].isin(selected_A)].copy()
    df_B_raw = filtered_df[filtered_df['Comp.SetDetail'].isin(selected_B)].copy()
    
    df_A_raw['Raw_Placement'] = df_A_raw['Rank'].apply(map_rank_to_placement)
    df_B_raw['Raw_Placement'] = df_B_raw['Rank'].apply(map_rank_to_placement)

    a_start = a_start_year if a_start_year is not None else 0
    b_start = b_start_year if b_start_year is not None else 0
    
    valid_a_cohort = df_A_raw[df_A_raw['Year_num'] >= a_start]['Person'].unique()

    def prepare_flow_data(df_a, df_b, a_mapping, b_mapping, a_order, b_order):
        df_a_mapped = df_a.copy()
        df_b_mapped = df_b.copy()

        df_a_mapped['Placement'] = df_a_mapped['Raw_Placement'].apply(a_mapping)
        df_b_mapped['Placement'] = df_b_mapped['Raw_Placement'].apply(b_mapping)

        df_a_mapped['Placement'] = pd.Categorical(df_a_mapped['Placement'].astype(str), categories=a_order, ordered=True)
        df_b_mapped['Placement'] = pd.Categorical(df_b_mapped['Placement'].astype(str), categories=b_order, ordered=True)

        df_a_mapped = df_a_mapped.sort_values(['Person', 'Placement', 'Year_num'])
        df_b_mapped = df_b_mapped.sort_values(['Person', 'Placement', 'Year_num'])

        best_a = df_a_mapped.drop_duplicates(subset=['Person'], keep='first')[['Person', 'Placement', 'Year_num']]
        best_a = best_a.set_index('Person').rename(columns={'Placement': 'A_Best', 'Year_num': 'A_Year'})

        best_b = df_b_mapped.drop_duplicates(subset=['Person'], keep='first')[['Person', 'Placement', 'Year_num']]
        best_b = best_b.set_index('Person').rename(columns={'Placement': 'B_Best', 'Year_num': 'B_Year'})

        master = pd.concat([best_a, best_b], axis=1)
        master['A_Best'] = master['A_Best'].astype(str).fillna('Geen deelname')
        master['B_Best'] = master['B_Best'].astype(str).fillna('Geen deelname')
        master['Conversion_Time'] = master['B_Year'] - master['A_Year']

        master.loc[master['Conversion_Time'] < 0, 'Conversion_Time'] = float('nan')
        master = master[~((master['A_Best'] == 'Geen deelname') & (master['B_Best'] == 'Geen deelname'))]
        return master

    # ------------------------------------------------------------------
    # 4. CONFIGUREER MAPPINGS
    # ------------------------------------------------------------------
    def map_to_podium(placement):
        if placement in ['1e', '2e', '3e']: return 'Podium'
        if placement == 'Geen deelname': return 'Geen deelname'
        return 'Geen Podium'

    def map_to_top8(placement):
        if placement in ['1e', '2e', '3e', '5e', '7e']: return 'Top 8'
        if placement == 'Geen deelname': return 'Geen deelname'
        return 'Geen Top 8'

    def map_identity(placement):
        return placement

    standard_order = ['1e', '2e', '3e', '5e', '7e', 'Deelname', 'Geen deelname']
    
    if grouping.lower() == 'podium':
        a_map = map_to_podium
        a_cats_orig = ['Podium', 'Geen Podium', 'Geen deelname']
    elif grouping.lower() == 'top8':
        a_map = map_to_top8
        a_cats_orig = ['Top 8', 'Geen Top 8', 'Geen deelname']
    else:
        a_map = map_identity
        a_cats_orig = standard_order

    if b_target.lower() == 'top8':
        b_target_name = "Top 8 Winnaars"
        b_target_placements = ['1e', '2e', '3e', '5e', '7e']
        sankey_b_placements = ['1e', '2e', '3e', '5e', '7e', 'Deelname']
        b_cats_conv = ['Top 8', 'Geen Top 8', 'Geen deelname']
        def b_map_conv(placement):
            if placement in ['1e', '2e', '3e', '5e', '7e']: return 'Top 8'
            if placement == 'Geen deelname': return 'Geen deelname'
            return 'Geen Top 8'
            
    elif b_target.lower() == 'podium':
        b_target_name = "Podium Winnaars"
        b_target_placements = ['1e', '2e', '3e']
        sankey_b_placements = ['1e', '2e', '3e'] 
        b_cats_conv = ['1e', '2e', '3e', 'Geen Podium', 'Geen deelname']
        def b_map_conv(placement):
            if placement in ['1e', '2e', '3e']: return placement
            if placement == 'Geen deelname': return 'Geen deelname'
            return 'Geen Podium'
            
    else: 
        b_target_name = "Top 3 Winnaars"
        b_target_placements = ['1e', '2e', '3e']
        sankey_b_placements = ['1e', '2e', '3e', '5e', '7e', 'Deelname']
        b_cats_conv = ['Top 3', 'Geen Top 3', 'Geen deelname']
        def b_map_conv(placement):
            if placement in ['1e', '2e', '3e']: return 'Top 3'
            if placement == 'Geen deelname': return 'Geen deelname'
            return 'Geen Top 3'

    valid_b_sankey_cohort = df_B_raw[
        (df_B_raw['Year_num'] >= b_start) & 
        (df_B_raw['Raw_Placement'].isin(sankey_b_placements))
    ]['Person'].unique()

    a_cats_conv = [c for c in a_cats_orig if c != 'Geen deelname']
    b_cats_orig = [c for c in b_cats_conv if c != 'Geen deelname']
    
    df_orig = prepare_flow_data(df_A_raw, df_B_raw, a_map, b_map_conv, a_cats_orig, b_cats_conv)
    df_orig = df_orig[df_orig.index.isin(valid_b_sankey_cohort)].copy()

    df_conv = prepare_flow_data(df_A_raw, df_B_raw, a_map, b_map_conv, a_cats_orig, b_cats_conv)
    df_conv = df_conv[df_conv.index.isin(valid_a_cohort)].copy()
    df_conv = df_conv[df_conv['A_Best'] != 'Geen deelname'].copy()

    a_title = ", ".join(selected_A).replace("World Cadet Championships", "WK Cadetten").replace("European Judo Championships Cadets", "EK Cadetten").replace("World Junior Championships", "WK Junioren").replace("European Judo Championships Juniors", "EK Junioren")
    b_title = ", ".join(selected_B).replace("World Senior Championships", "WK Senioren").replace("Olympic Games", "Olympische Spelen")
    filter_string = f"<br>Filters: [{', '.join(title_filters)}]" if title_filters else "<br>Filters: [Alle Atleten]"

# ------------------------------------------------------------------
    # 5. INTERNE HELPER: RENDER SANKEY (MET HTML EXPORT EN KLEUREN)
    # ------------------------------------------------------------------
    def _plot_sankey(df_plot, a_cats, b_cats, chart_title, calc_direction='forward', a_name='A', b_name='B'):
        df_plot['A_Best'] = pd.Categorical(df_plot['A_Best'].astype(str), categories=a_cats, ordered=True)
        df_plot['B_Best'] = pd.Categorical(df_plot['B_Best'].astype(str), categories=b_cats, ordered=True)
        df_plot = df_plot.dropna(subset=['A_Best', 'B_Best'])
        
        flow_grouped = df_plot.groupby(['A_Best', 'B_Best'], observed=False)
        flow_counts = flow_grouped.size().reset_index(name='Count')
        flow_times = flow_grouped['Conversion_Time'].mean().reset_index(name='Avg_Time')
        
        flow_counts = pd.merge(flow_counts, flow_times, on=['A_Best', 'B_Best'])
        flow_counts = flow_counts[flow_counts['Count'] > 0]
        
        if flow_counts.empty:
            print(f"Geen data beschikbaar voor: {chart_title}")
            return
            
        # --- DYNAMISCHE TEKST EN PIJLEN ---
        if calc_direction == 'forward':
            flow_counts['Percentage'] = flow_counts.groupby('A_Best')['Count'].transform(lambda x: (x / x.sum()) * 100)
            hover_prefix = "Conversie"
            hover_template = "%{source.label} → %{target.label}<br>Aantal Atleten: %{value}<br>%{customdata}<extra></extra>"
        else:
            flow_counts['Percentage'] = flow_counts.groupby('B_Best')['Count'].transform(lambda x: (x / x.sum()) * 100)
            hover_prefix = "Herkomst"
            hover_template = "%{target.label} ← %{source.label}<br>Aantal Atleten: %{value}<br>%{customdata}<extra></extra>"
            
        flow_counts['Hover_Custom'] = flow_counts.apply(
            lambda row: f"{hover_prefix}: {row['Percentage']:.1f}%<br>Gem. Tijd: {row['Avg_Time']:.1f} jr" if pd.notna(row['Avg_Time']) else f"{hover_prefix}: {row['Percentage']:.1f}%<br>Gem. Tijd: N.v.t.", axis=1
        )

        a_suffix = f" ({a_name})"
        b_suffix = f" ({b_name})"
        
        flow_counts['Source_Node'] = flow_counts['A_Best'].astype(str) + a_suffix
        flow_counts['Target_Node'] = flow_counts['B_Best'].astype(str) + b_suffix

        a_nodes = sorted(flow_counts['Source_Node'].unique(), key=lambda n: a_cats.index(n.replace(a_suffix, '')))
        b_nodes = sorted(flow_counts['Target_Node'].unique(), key=lambda n: b_cats.index(n.replace(b_suffix, '')))
        
        all_nodes = a_nodes + b_nodes
        node_dict = {node: i for i, node in enumerate(all_nodes)}
        flow_counts['source_idx'] = flow_counts['Source_Node'].map(node_dict)
        flow_counts['target_idx'] = flow_counts['Target_Node'].map(node_dict)

        a_volumes = flow_counts.groupby('Source_Node')['Count'].sum()
        b_volumes = flow_counts.groupby('Target_Node')['Count'].sum()
        total_volumes = {**a_volumes.to_dict(), **b_volumes.to_dict()}
        
        # --- KLEUREN HELPER ---
        def get_color(name, is_link=False):
            alpha = 0.4 if is_link else 0.9
            name_lower = name.lower()
            
            if 'geen' in name_lower or 'not' in name_lower:
                return f"rgba(210, 210, 210, {0.35 if is_link else 0.7})"
            elif any(x in name_lower for x in ['1e', '2e', '3e', 'podium', 'top 3']):
                return f"rgba(255, 102, 0, {alpha})"  
            elif any(x in name_lower for x in ['5e', '7e', 'top 8']):
                return f"rgba(0, 55, 138, {alpha})"   
            else:
                return f"rgba(130, 140, 150, {alpha})" 

        display_node_labels = []
        node_colors = []
        
        def abbrev_name(name):
            name = name.replace(b_suffix, '').replace(a_suffix, '')
            if name == 'Geen deelname': return 'Geen Deelname'
            if name == 'Deelname': return 'Deelname'
            return name
            
        for node in all_nodes:
            clean_name = abbrev_name(node)
            label = f"<b>{clean_name} (n={total_volumes.get(node, 0)})</b>"
            display_node_labels.append(label)
            node_colors.append(get_color(clean_name, is_link=False))

        link_colors = [get_color(abbrev_name(src), is_link=True) for src in flow_counts['Source_Node']]

        node_x, node_y = [], []
        total_a_vol, total_b_vol = a_volumes.sum(), b_volumes.sum()
        usable_height = 1 
        
        current_y = 0.01
        for node in a_nodes:
            proportion = (a_volumes.get(node, 0) / total_a_vol) * usable_height
            node_x.append(0.01)
            node_y.append(current_y + (proportion / 2))
            current_y += proportion

        current_y = 0.01
        for node in b_nodes:
            proportion = (b_volumes.get(node, 0) / total_b_vol) * usable_height
            node_x.append(0.99)
            node_y.append(current_y + (proportion / 2))
            current_y += proportion

        layout_margin = dict(t=120, l=165, r=50) if calc_direction == 'forward' else dict(t=120, l=50, r=165)

        fig = go.Figure(data=[go.Sankey(
            arrangement="snap",
            node=dict(
                pad=15, thickness=25, line=dict(color="white", width=1),
                label=display_node_labels, x=node_x, y=node_y,
                color=node_colors, 
                hovertemplate="%{label}<extra></extra>"
            ),
            link=dict(
                source=flow_counts['source_idx'], target=flow_counts['target_idx'],
                value=flow_counts['Count'], customdata=flow_counts['Hover_Custom'],
                hovertemplate=hover_template, # DYNAMISCH OP BASIS VAN DIRECTIE
                color=link_colors 
            )
        )])
        
        fig.update_layout(
            title_text=chart_title, 
            font=dict(size=11, family="HK Grotesk, Open Sans, sans-serif"), 
            height=1000, 
            margin=layout_margin
        )
        
        if export_html_path:
            fig.write_html(export_html_path)
            print(f"Sankey succesvol opgeslagen naar {export_html_path}")

        fig.show()

    # ------------------------------------------------------------------
    # 6. RENDER HET DIAGRAM
    # ------------------------------------------------------------------
    if plot_type.lower() in ['both', 'origination']:
        df_raw = prepare_flow_data(df_A_raw, df_B_raw, map_identity, map_identity, standard_order, standard_order)
        valid_b_winners = df_B_raw[(df_B_raw['Year_num'] >= b_start) & (df_B_raw['Raw_Placement'].isin(b_target_placements))]['Person'].unique()
        df_raw = df_raw[df_raw.index.isin(valid_b_winners)]
        
        title_orig = f"<b>Herkomst Flowdiagram</b><br>Basis: {a_title}<br>Doel ({b_target_name}): {b_title}{filter_string}"
        _plot_sankey(df_orig, a_cats_orig, b_cats_orig, title_orig, calc_direction='backward', a_name=a_title, b_name=b_title)

    if plot_type.lower() in ['both', 'conversion']:
        title_conv = f"<b>Conversie Flowdiagram</b><br>Basis (Alleen Deelnemers): {a_title}<br>Doel ({b_target_name}): {b_title}{filter_string}"
        _plot_sankey(df_conv, a_cats_conv, b_cats_conv, title_conv, calc_direction='forward', a_name=a_title, b_name=b_title)

    return
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

def plot_european_youth_conversion_bar_chart(
    df, 
    start_year=2016, 
    gender='All', 
    european_only=True, 
    u18_events=None,
    u21_events=None
):
    """
    Creates a grouped bar chart showing the percentage of Senior Medallists 
    who previously achieved medals, Top 5s, Top 7s, and participations.
    Embeds a single interactive Dropdown Menu directly into the HTML to toggle views.
    Vertaling naar Nederlands en NOC*NSF huisstijl toegepast.
    """
    
    # Standaard toernooien instellen indien niet opgegeven
    if u18_events is None:
        u18_events = ['European Judo Championships Cadets', 'EYOF', 'World Cadet Championships']
    if u21_events is None:
        u21_events = ['European Judo Championships Juniors', 'World Junior Championships']

    df_clean = df.copy()
    df_clean['Year_num'] = pd.to_numeric(df_clean['Year'], errors='coerce')

    # 1. Standaardiseer rangschikking naar niveaus (tiers)
    def get_rank_level(rank_val):
        if pd.isna(rank_val): return 999
        r = str(rank_val).strip().lower()
        if r in ['nan', 'none', '']: return 999
        if r in ['1', '1.0', '1st', '2', '2.0', '2nd', '3', '3.0', '3rd']: return 3
        if r in ['5', '5.0', '5th']: return 5
        if r in ['7', '7.0', '7th']: return 7
        return 99 # Elke andere rank betekent algemene deelname

    df_clean['Rank_Level'] = df_clean['Rank'].apply(get_rank_level)
    df_clean['Is_Medallist'] = df_clean['Rank_Level'] <= 3

    # 2. Basis filters voor Senioren Doeltoernooien
    senior_tournaments = ['Olympic Games', 'World Senior Championships']
    senior_df = df_clean[
        (df_clean['Comp.SetDetail'].isin(senior_tournaments)) & 
        (df_clean['Year_num'] >= start_year) & 
        (df_clean['Is_Medallist'])
    ].copy()

    title_filters = []

    if gender and gender.lower() != 'all':
        senior_df = senior_df[senior_df['Gender'].astype(str).str.lower() == gender.lower()]
        nl_gender = "Mannen" if gender.lower() in ['m', 'men', 'male'] else ("Vrouwen" if gender.lower() in ['f', 'w', 'women', 'female'] else gender)
        title_filters.append(f"Geslacht: {nl_gender}")

    if european_only:
        senior_df = senior_df[senior_df['Continent'].astype(str).str.lower() == 'europe']
        title_filters.append("Alleen Europese Medaillewinnaars")

    # 3. Data-extractie o.b.v. opgegeven evenementen
    def get_tier_years(event_list, max_rank):
        subset = df_clean[(df_clean['Comp.SetDetail'].isin(event_list)) & (df_clean['Rank_Level'] <= max_rank)]
        return subset.groupby('Person')['Year_num'].min().to_dict()

    u18_comb_t3 = get_tier_years(u18_events, 3)
    u18_comb_t5 = get_tier_years(u18_events, 5)
    u18_comb_t7 = get_tier_years(u18_events, 7)
    u18_comb_part = get_tier_years(u18_events, 99)

    u21_comb_t3 = get_tier_years(u21_events, 3)
    u21_comb_t5 = get_tier_years(u21_events, 5)
    u21_comb_t7 = get_tier_years(u21_events, 7)
    u21_comb_part = get_tier_years(u21_events, 99)

    # 4. Genereer Evenement Labels en sorteer chronologisch
    senior_df['Event_Label'] = senior_df['Year_num'].astype(int).astype(str) + " " + senior_df['Comp.SetDetail']
    unique_events = senior_df[['Year_num', 'Comp.SetDetail', 'Event_Label']].drop_duplicates()

    event_order = {'World Senior Championships': 1, 'Olympic Games': 2}
    unique_events['Order'] = unique_events['Comp.SetDetail'].map(event_order).fillna(3)
    unique_events = unique_events.sort_values(by=['Year_num', 'Order', 'Comp.SetDetail'])

    # 5. Bereken percentages en bouw lijsten van atleten
    results = []

    def format_hover_list(athlete_list, chunk_size=3):
        if not athlete_list: return " Geen"
        chunks = [", ".join(athlete_list[i:i+chunk_size]) for i in range(0, len(athlete_list), chunk_size)]
        return "<br>  " + "<br>  ".join(chunks)

    for _, row in unique_events.iterrows():
        event_label = row['Event_Label']
        event_year = row['Year_num']

        event_medallists = senior_df[senior_df['Event_Label'] == event_label]['Person'].dropna().unique()
        total_medallists = len(event_medallists)

        if total_medallists == 0:
            continue

        def count_achievers(target_dict):
            return sum(1 for p in event_medallists if p in target_dict and target_dict[p] <= event_year)

        u18_t3_list, u18_t5_list, u18_t7_list, u18_part_list = [], [], [], []
        u21_t3_list, u21_t5_list, u21_t7_list, u21_part_list = [], [], [], []
        
        for p in event_medallists:
            if p in u18_comb_t3 and u18_comb_t3[p] <= event_year: u18_t3_list.append(f"{p} ({int(u18_comb_t3[p])})")
            elif p in u18_comb_t5 and u18_comb_t5[p] <= event_year: u18_t5_list.append(f"{p} ({int(u18_comb_t5[p])})")
            elif p in u18_comb_t7 and u18_comb_t7[p] <= event_year: u18_t7_list.append(f"{p} ({int(u18_comb_t7[p])})")
            elif p in u18_comb_part and u18_comb_part[p] <= event_year: u18_part_list.append(f"{p} ({int(u18_comb_part[p])})")

            if p in u21_comb_t3 and u21_comb_t3[p] <= event_year: u21_t3_list.append(f"{p} ({int(u21_comb_t3[p])})")
            elif p in u21_comb_t5 and u21_comb_t5[p] <= event_year: u21_t5_list.append(f"{p} ({int(u21_comb_t5[p])})")
            elif p in u21_comb_t7 and u21_comb_t7[p] <= event_year: u21_t7_list.append(f"{p} ({int(u21_comb_t7[p])})")
            elif p in u21_comb_part and u21_comb_part[p] <= event_year: u21_part_list.append(f"{p} ({int(u21_comb_part[p])})")

        results.append({
            'Event': event_label.replace("Olympic Games", "Olympische Spelen").replace("World Senior Championships", "WK Senioren"),
            'Year': event_year,
            'Total_Medallists': total_medallists,

            'U18_Comb_T3_Pct': count_achievers(u18_comb_t3) / total_medallists * 100,
            'U18_Comb_T5_Pct': count_achievers(u18_comb_t5) / total_medallists * 100,
            'U18_Comb_T7_Pct': count_achievers(u18_comb_t7) / total_medallists * 100,
            'U18_Comb_Part_Pct': count_achievers(u18_comb_part) / total_medallists * 100,
            'U18_Comb_T3_Count': count_achievers(u18_comb_t3),
            'U18_Comb_T5_Count': count_achievers(u18_comb_t5),
            'U18_Comb_T7_Count': count_achievers(u18_comb_t7),
            'U18_Comb_Part_Count': count_achievers(u18_comb_part),
            'U18_Comb_T3_Hover': format_hover_list(u18_t3_list),
            'U18_Comb_T5_Hover': format_hover_list(u18_t5_list),
            'U18_Comb_T7_Hover': format_hover_list(u18_t7_list),
            'U18_Comb_Part_Hover': format_hover_list(u18_part_list),

            'U21_Comb_T3_Pct': count_achievers(u21_comb_t3) / total_medallists * 100,
            'U21_Comb_T5_Pct': count_achievers(u21_comb_t5) / total_medallists * 100,
            'U21_Comb_T7_Pct': count_achievers(u21_comb_t7) / total_medallists * 100,
            'U21_Comb_Part_Pct': count_achievers(u21_comb_part) / total_medallists * 100,
            'U21_Comb_T3_Count': count_achievers(u21_comb_t3),
            'U21_Comb_T5_Count': count_achievers(u21_comb_t5),
            'U21_Comb_T7_Count': count_achievers(u21_comb_t7),
            'U21_Comb_Part_Count': count_achievers(u21_comb_part),
            'U21_Comb_T3_Hover': format_hover_list(u21_t3_list),
            'U21_Comb_T5_Hover': format_hover_list(u21_t5_list),
            'U21_Comb_T7_Hover': format_hover_list(u21_t7_list),
            'U21_Comb_Part_Hover': format_hover_list(u21_part_list),
        })

    res_df = pd.DataFrame(results)

    if res_df.empty:
        print("Geen data beschikbaar voor deze specifieke filters.")
        return None, res_df

    filter_string = f"[{', '.join(title_filters)}]" if title_filters else "[Allemaal]"
    u18_title_str = ", ".join(u18_events)
    u21_title_str = ", ".join(u21_events)

    # 6. Render Groepering Bar Chart
    fig = go.Figure()

    def get_tier_texts(prefix):
        t3_txt, t5_txt, t7_txt, part_txt = [], [], [], []
        for _, r in res_df.iterrows():
            t3 = r[f'{prefix}_T3_Pct']; t5 = r[f'{prefix}_T5_Pct']
            t7 = r[f'{prefix}_T7_Pct']; p = r[f'{prefix}_Part_Pct']
            
            t3_txt.append(f"<b>T3: {t3:.0f}%</b>" if t3 > 0 else "")
            t5_txt.append(f"<b>T5: {t5:.0f}%</b>" if t5 - t3 >= 2 else "")
            t7_txt.append(f"<b>T7: {t7:.0f}%</b>" if t7 - t5 >= 2 else "")
            part_txt.append(f"<b>D: {p:.0f}%</b>" if p - t7 >= 2 else "") # D = Deelname
        return t3_txt, t5_txt, t7_txt, part_txt

    def build_hover(prefix, tier_name, hover_idx):
        return (
            "<b>%{x}</b><br>"
            f"Top 3: %{{customdata[0]}} / %{{customdata[1]}} (%{{customdata[8]:.1f}}%)<br>"
            f"Top 5: %{{customdata[2]}} / %{{customdata[1]}} (%{{customdata[3]:.1f}}%)<br>"
            f"Top 7: %{{customdata[4]}} / %{{customdata[1]}} (%{{customdata[5]:.1f}}%)<br>"
            f"Deelname: %{{customdata[6]}} / %{{customdata[1]}} (%{{customdata[7]:.1f}}%)<br><br>"
            f"<b>{prefix} {tier_name} (Vroegste Jaar):</b>%{{customdata[{hover_idx}]}}"
            "<extra></extra>"
        )

    def get_customdata(prefix):
        return np.column_stack((
            res_df[f'{prefix}_T3_Count'], res_df['Total_Medallists'],      # 0, 1
            res_df[f'{prefix}_T5_Count'], res_df[f'{prefix}_T5_Pct'],      # 2, 3
            res_df[f'{prefix}_T7_Count'], res_df[f'{prefix}_T7_Pct'],      # 4, 5
            res_df[f'{prefix}_Part_Count'], res_df[f'{prefix}_Part_Pct'],  # 6, 7
            res_df[f'{prefix}_T3_Pct'],                                    # 8
            res_df[f'{prefix}_T3_Hover'],                                  # 9
            res_df[f'{prefix}_T5_Hover'],                                  # 10
            res_df[f'{prefix}_T7_Hover'],                                  # 11
            res_df[f'{prefix}_Part_Hover']                                 # 12
        ))

    def add_group_traces(prefix, offset_name, legend_name, rgb_str):
        customdata = get_customdata(prefix)
        t3_txt, t5_txt, t7_txt, part_txt = get_tier_texts(prefix)
        uniform_font = dict(size=11, color='white', family="HK Grotesk, Open Sans, sans-serif")

        # Trace Index mapping: U18 (0:Part, 1:T7, 2:T5, 3:T3), U21 (4:Part, 5:T7, 6:T5, 7:T3)
        
        # 1. Deelname (Tallest, Base Layer)
        fig.add_trace(go.Bar(
            x=res_df['Event'], y=res_df[f'{prefix}_Part_Pct'],
            offsetgroup=offset_name, legendgroup=offset_name,
            name=f'Herkomst {legend_name}',
            marker=dict(color=f'rgba({rgb_str}, 0.25)', line=dict(width=0)),
            text=part_txt, textposition='inside', insidetextanchor='end', 
            textangle=0, textfont=uniform_font, 
            customdata=customdata, hovertemplate=build_hover(legend_name, "Alleen Deelname", 12), 
            showlegend=True 
        ))
        
        # 2. Top 7 
        fig.add_trace(go.Bar(
            x=res_df['Event'], y=res_df[f'{prefix}_T7_Pct'],
            offsetgroup=offset_name, legendgroup=offset_name,
            name=f'Herkomst {legend_name}',
            marker=dict(color=f'rgba({rgb_str}, 0.45)', line=dict(width=0)),
            text=t7_txt, textposition='inside', insidetextanchor='end', 
            textangle=0, textfont=uniform_font,
            customdata=customdata, hovertemplate=build_hover(legend_name, "Alleen Top 7", 11), 
            showlegend=False
        ))
        
        # 3. Top 5 
        fig.add_trace(go.Bar(
            x=res_df['Event'], y=res_df[f'{prefix}_T5_Pct'],
            offsetgroup=offset_name, legendgroup=offset_name,
            name=f'Herkomst {legend_name}',
            marker=dict(color=f'rgba({rgb_str}, 0.70)', line=dict(width=0)),
            text=t5_txt, textposition='inside', insidetextanchor='end', 
            textangle=0, textfont=uniform_font,
            customdata=customdata, hovertemplate=build_hover(legend_name, "Alleen Top 5", 10), 
            showlegend=False
        ))
        
        # 4. Top 3 (Kortste, Bovenste Laag)
        fig.add_trace(go.Bar(
            x=res_df['Event'], y=res_df[f'{prefix}_T3_Pct'],
            offsetgroup=offset_name, legendgroup=offset_name,
            name=f'Herkomst {legend_name}',
            marker=dict(color=f'rgba({rgb_str}, 1.0)', line=dict(width=0)),
            text=t3_txt, textposition='inside', insidetextanchor='end', 
            textangle=0, textfont=uniform_font,
            customdata=customdata, hovertemplate=build_hover(legend_name, "Top 3 (Medaille)", 9), 
            showlegend=False
        ))

    # Pas NOC*NSF kleuren toe: Oranje voor U18, Blauw voor U21
    add_group_traces('U18_Comb', 'u18', 'U18', '255, 102, 0') # Oranje
    add_group_traces('U21_Comb', 'u21', 'U21', '0, 55, 138')  # Blauw

    # --- HTML INTERACTIVE SINGLE DROPDOWN ---
    dropdown_buttons = [
        dict(label="Weergave: Alle Niveaus (Standaard)",
             method="restyle",
             args=[{"visible": [True, True, True, True, True, True, True, True],
                    "showlegend": [True, False, False, False, True, False, False, False]}]),
                    
        dict(label="Weergave: Alleen Medailles (Top 3)",
             method="restyle",
             args=[{"visible": [False, False, False, True, False, False, False, True],
                    "showlegend": [False, False, False, True, False, False, False, True]}]),

        dict(label="Weergave: Medailles & Top 5",
             method="restyle",
             args=[{"visible": [False, False, True, True, False, False, True, True],
                    "showlegend": [False, False, True, False, False, False, True, False]}]),

        dict(label="Weergave: Medailles, Top 5 & Top 7",
             method="restyle",
             args=[{"visible": [False, True, True, True, False, True, True, True],
                    "showlegend": [False, True, False, False, False, True, False, False]}]),
                    
        dict(label="Weergave: Medailles, Top 7 & Deelname",
             method="restyle",
             args=[{"visible": [True, True, False, True, True, True, False, True],
                    "showlegend": [True, False, False, False, True, False, False, False]}]),
                    
        dict(label="Weergave: Medailles & Deelname",
             method="restyle",
             args=[{"visible": [True, False, False, True, True, False, False, True],
                    "showlegend": [True, False, False, False, True, False, False, False]}]),
                    
        dict(label="Weergave: Alleen Deelname",
             method="restyle",
             args=[{"visible": [True, False, False, False, True, False, False, False],
                    "showlegend": [True, False, False, False, True, False, False, False]}])
    ]

    fig.update_layout(
        title=dict(
            text=(
                f"<b>Herkomst Senioren Medaillewinnaars & Eerdere Jeugdklasseringen (Sinds {start_year})</b><br>"
                f"<span style='font-size:12px; color:gray'>U18 Toernooien: {u18_title_str} | U21 Toernooien: {u21_title_str}<br>"
                f"Filters: {filter_string}</span>"
            ),
            y=0.95,
            x=0.01,
            xanchor='left',
            yanchor='top',
            font=dict(family="HK Grotesk, Open Sans, sans-serif")
        ),
        xaxis_title="Senioren Toernooi",
        yaxis_title="Percentage van Senioren Medaillewinnaars (%)",
        barmode='group',
        yaxis=dict(range=[0, 105]), 
        margin=dict(t=180), 
        font=dict(family="HK Grotesk, Open Sans, sans-serif", size=11),
        height=750,
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="right",
            x=1
        ),
        updatemenus=[
            dict(
                active=0,
                buttons=dropdown_buttons,
                direction="down",
                pad={"r": 10, "t": 10},
                showactive=True,
                x=0.01,
                xanchor="left",
                y=1.15,
                yanchor="top",
                font=dict(size=12, color="black", family="HK Grotesk, Open Sans, sans-serif"),
                bgcolor="white",
                bordercolor="gray",
                borderwidth=1
            )
        ]
    )

    fig.show()

    return fig, res_df
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

def plot_forward_conversion_bar_chart(
    df, 
    cadet_events=None, 
    senior_events=None, 
    start_year=2009,
    end_year=2024,
    gender='All', 
    european_only=False,
    dutch_only=False
):
    """
    Creates a stacked grouped bar chart showing FORWARD CONVERSION.
    X-Axis: Specific Cadet/Youth Tournaments by Year.
    Y-Axis: The percentage of Medallists from that specific tournament 
            who went on to achieve T3, T5, T7, or Participation at Senior targets.
    Hovering reveals the specific athletes who peaked at each Senior tier.
    Volledig vertaald naar het Nederlands en NOC*NSF huisstijl.
    """
    
    if cadet_events is None:
        cadet_events = ['World Cadet Championships']
    if senior_events is None:
        senior_events = ['World Senior Championships', 'Olympic Games']

    df_clean = df.copy()
    df_clean['Year_num'] = pd.to_numeric(df_clean['Year'], errors='coerce')

    # 1. Standaardiseer rangschikking naar niveaus (tiers)
    def get_rank_level(rank_val):
        if pd.isna(rank_val): return 999
        r = str(rank_val).strip().lower()
        if r in ['nan', 'none', '']: return 999
        if r in ['1', '1.0', '1st', '2', '2.0', '2nd', '3', '3.0', '3rd']: return 3
        if r in ['5', '5.0', '5th']: return 5
        if r in ['7', '7.0', '7th']: return 7
        return 99 # Elke andere rank betekent algemene deelname

    df_clean['Rank_Level'] = df_clean['Rank'].apply(get_rank_level)

    # 2. Demografische Filters
    title_filters = []

    if gender and gender.lower() != 'all':
        df_clean = df_clean[df_clean['Gender'].astype(str).str.lower() == gender.lower()]
        nl_gender = "Mannen" if gender.lower() in ['m', 'men', 'male'] else ("Vrouwen" if gender.lower() in ['f', 'w', 'women', 'female'] else gender)
        title_filters.append(f"Geslacht: {nl_gender}")

    if dutch_only:
        dutch_mask = (
            df_clean['Country'].astype(str).str.lower().isin(['netherlands', 'the netherlands', 'ned']) |
            (df_clean['Country Code'].astype(str).str.upper() == 'NED') |
            df_clean['Nationality'].astype(str).str.upper().isin(['NED', 'NETHERLANDS', 'DUTCH'])
        )
        df_clean = df_clean[df_clean['Person'].isin(df_clean[dutch_mask]['Person'].unique())]
        title_filters.append("Nederlandse Atleten")
    elif european_only:
        eju_codes = ['ALB', 'AND', 'ARM', 'AUT', 'AZE', 'BLR', 'BEL', 'BIH', 'BUL', 'CRO', 'CYP', 'CZE', 
                     'DEN', 'EST', 'FIN', 'FRA', 'GEO', 'GER', 'GBR', 'GRE', 'HUN', 'ISL', 'IRL', 'ISR', 
                     'ITA', 'KOS', 'LAT', 'LIE', 'LTU', 'LUX', 'MLT', 'MDA', 'MON', 'MNE', 'NED', 
                     'MKD', 'NOR', 'POL', 'POR', 'ROU', 'RUS', 'SMR', 'SRB', 'SVK', 'SLO', 'ESP', 
                     'SWE', 'SUI', 'TUR', 'UKR']
        euro_mask = (df_clean['Continent'].astype(str).str.lower() == 'europe') | \
                    (df_clean['Country Code'].astype(str).str.upper().isin(eju_codes)) | \
                    (df_clean['Nationality'].astype(str).str.upper().isin(eju_codes))
        df_clean = df_clean[df_clean['Person'].isin(df_clean[euro_mask]['Person'].unique())]
        title_filters.append("Europese Atleten")

    # 3. Identificeer Basis Cohorts (Jeugd Medaillewinnaars)
    cadet_df = df_clean[
        (df_clean['Comp.SetDetail'].isin(cadet_events)) & 
        (df_clean['Rank_Level'] <= 3) & # MOET een jeugd medaillewinnaar zijn
        (df_clean['Year_num'] >= start_year) &
        (df_clean['Year_num'] <= end_year)
    ].copy()

    # Haal unieke jaren/evenementen op voor de X-as
    cadet_df['Event_Label'] = cadet_df['Year_num'].astype(int).astype(str) + " " + cadet_df['Comp.SetDetail']
    unique_events = cadet_df[['Year_num', 'Comp.SetDetail', 'Event_Label']].drop_duplicates()
    unique_events = unique_events.sort_values(by=['Year_num', 'Comp.SetDetail'])

    # 4. Extraheer Senioren Doelhistorie
    senior_df = df_clean[df_clean['Comp.SetDetail'].isin(senior_events)].copy()

    # 5. Verwerk Berekeningen
    results = []
    
    def format_hover_list(athlete_list, chunk_size=3):
        if not athlete_list: return " Geen"
        chunks = [", ".join(athlete_list[i:i+chunk_size]) for i in range(0, len(athlete_list), chunk_size)]
        return "<br>  " + "<br>  ".join(chunks)

    for _, row in unique_events.iterrows():
        c_year = row['Year_num']
        c_event = row['Comp.SetDetail']
        event_label = row['Event_Label']

        cohort_persons = cadet_df[
            (cadet_df['Year_num'] == c_year) & 
            (cadet_df['Comp.SetDetail'] == c_event)
        ]['Person'].unique()

        total_cohort = len(cohort_persons)
        if total_cohort == 0: continue

        senior_results = senior_df[
            (senior_df['Person'].isin(cohort_persons)) & 
            (senior_df['Year_num'] >= c_year)
        ]

        if not senior_results.empty:
            idx = senior_results.groupby('Person')['Rank_Level'].idxmin()
            best_senior = senior_results.loc[idx].set_index('Person')
        else:
            best_senior = pd.DataFrame(columns=['Rank_Level', 'Year_num'])

        t3_count, t5_count, t7_count, part_count = 0, 0, 0, 0
        t3_list, t5_list, t7_list, part_list = [], [], [], []

        for p in cohort_persons:
            if p in best_senior.index:
                rank = best_senior.loc[p, 'Rank_Level']
                yr = int(best_senior.loc[p, 'Year_num'])
                
                if rank <= 3: 
                    t3_count += 1; t5_count += 1; t7_count += 1; part_count += 1
                    t3_list.append(f"{p} ({yr})")
                elif rank <= 5: 
                    t5_count += 1; t7_count += 1; part_count += 1
                    t5_list.append(f"{p} ({yr})")
                elif rank <= 7: 
                    t7_count += 1; part_count += 1
                    t7_list.append(f"{p} ({yr})")
                elif rank <= 99: 
                    part_count += 1
                    part_list.append(f"{p} ({yr})")

        # Vertaal evenement labels voor de grafiek weergave (indien gewenst)
        nl_event_label = event_label.replace("World Cadet Championships", "WK Cadetten").replace("European Judo Championships Cadets", "EK Cadetten")

        results.append({
            'Event_Label': nl_event_label,
            'Total_Cohort': total_cohort,
            'T3_Count': t3_count, 'T3_Pct': t3_count / total_cohort * 100,
            'T5_Count': t5_count, 'T5_Pct': t5_count / total_cohort * 100,
            'T7_Count': t7_count, 'T7_Pct': t7_count / total_cohort * 100,
            'Part_Count': part_count, 'Part_Pct': part_count / total_cohort * 100,
            'T3_Hover': format_hover_list(t3_list),
            'T5_Hover': format_hover_list(t5_list),
            'T7_Hover': format_hover_list(t7_list),
            'Part_Hover': format_hover_list(part_list)
        })

    res_df = pd.DataFrame(results)

    # --- CRITICAL FIX: Safe unpack return for empty dataframes ---
    if res_df.empty:
        print("Geen data beschikbaar voor deze specifieke filters.")
        return None, res_df

    cadet_str = ", ".join(cadet_events).replace("World Cadet Championships", "WK Cadetten").replace("European Judo Championships Cadets", "EK Cadetten")
    senior_str = ", ".join(senior_events).replace("Olympic Games", "Olympische Spelen").replace("World Senior Championships", "WK Senioren")
    filter_string = f"[{', '.join(title_filters)}]" if title_filters else "[Allemaal]"

    # 6. Render Grouped Bar Chart
    fig = go.Figure()

    def get_tier_texts():
        t3_txt, t5_txt, t7_txt, part_txt = [], [], [], []
        for _, r in res_df.iterrows():
            t3 = r['T3_Pct']; t5 = r['T5_Pct']; t7 = r['T7_Pct']; p = r['Part_Pct']
            
            t3_txt.append(f"<b>T3: {t3:.0f}%</b>" if t3 > 0 else "")
            t5_txt.append(f"<b>T5: {t5:.0f}%</b>" if t5 - t3 >= 2 else "")
            t7_txt.append(f"<b>T7: {t7:.0f}%</b>" if t7 - t5 >= 2 else "")
            part_txt.append(f"<b>D: {p:.0f}%</b>" if p - t7 >= 2 else "") # D = Deelname
        return t3_txt, t5_txt, t7_txt, part_txt

    t3_txt, t5_txt, t7_txt, part_txt = get_tier_texts()

    def build_hover(tier_name, hover_idx):
        return (
            "<b>%{x}</b><br>"
            f"Top 3: %{{customdata[0]}} / %{{customdata[1]}} (%{{customdata[8]:.1f}}%)<br>"
            f"Top 5: %{{customdata[2]}} / %{{customdata[1]}} (%{{customdata[3]:.1f}}%)<br>"
            f"Top 7: %{{customdata[4]}} / %{{customdata[1]}} (%{{customdata[5]:.1f}}%)<br>"
            f"Deelname: %{{customdata[6]}} / %{{customdata[1]}} (%{{customdata[7]:.1f}}%)<br><br>"
            f"<b>Senioren {tier_name} (Vroegste Jaar):</b>%{{customdata[{hover_idx}]}}"
            "<extra></extra>"
        )

    customdata = np.column_stack((
        res_df['T3_Count'], res_df['Total_Cohort'],      
        res_df['T5_Count'], res_df['T5_Pct'],            
        res_df['T7_Count'], res_df['T7_Pct'],            
        res_df['Part_Count'], res_df['Part_Pct'],        
        res_df['T3_Pct'],                                
        res_df['T3_Hover'],                              
        res_df['T5_Hover'],                              
        res_df['T7_Hover'],                              
        res_df['Part_Hover']                             
    ))

    uniform_font = dict(size=12, color='white', family="HK Grotesk, Open Sans, sans-serif")
    rgb_str = '255, 102, 0' # NOC*NSF Oranje

    fig.add_trace(go.Bar(
        x=res_df['Event_Label'], y=res_df['Part_Pct'],
        offsetgroup='A', name='Deelname',
        marker=dict(color=f'rgba({rgb_str}, 0.25)', line=dict(width=0)),
        text=part_txt, textposition='inside', insidetextanchor='end', 
        textangle=0, textfont=uniform_font, 
        customdata=customdata, hovertemplate=build_hover('Alleen Deelname', 12), showlegend=True
    ))
    
    fig.add_trace(go.Bar(
        x=res_df['Event_Label'], y=res_df['T7_Pct'],
        offsetgroup='A', name='Top 7',
        marker=dict(color=f'rgba({rgb_str}, 0.45)', line=dict(width=0)),
        text=t7_txt, textposition='inside', insidetextanchor='end', 
        textangle=0, textfont=uniform_font,
        customdata=customdata, hovertemplate=build_hover('Alleen Top 7', 11), showlegend=True
    ))
    
    fig.add_trace(go.Bar(
        x=res_df['Event_Label'], y=res_df['T5_Pct'],
        offsetgroup='A', name='Top 5',
        marker=dict(color=f'rgba({rgb_str}, 0.70)', line=dict(width=0)),
        text=t5_txt, textposition='inside', insidetextanchor='end', 
        textangle=0, textfont=uniform_font,
        customdata=customdata, hovertemplate=build_hover('Alleen Top 5', 10), showlegend=True
    ))
    
    fig.add_trace(go.Bar(
        x=res_df['Event_Label'], y=res_df['T3_Pct'],
        offsetgroup='A', name='Top 3 (Medaille)',
        marker=dict(color=f'rgba({rgb_str}, 1.0)', line=dict(width=0)),
        text=t3_txt, textposition='inside', insidetextanchor='end', 
        textangle=0, textfont=uniform_font,
        customdata=customdata, hovertemplate=build_hover('Top 3 (Medaille)', 9), showlegend=True
    ))

    fig.update_layout(
        title=dict(
            text=(
                f"<b>Conversie: {cadet_str} Medaillewinnaars naar Senioren Prestatie</b><br>"
                f"<span style='font-size:12px; color:gray'>Doeltoernooien: {senior_str}<br>"
                f"Filters: {filter_string} | Jaren: {start_year}-{end_year}</span>"
            ),
            font=dict(family="HK Grotesk, Open Sans, sans-serif")
        ),
        xaxis_title="Toernooi & Jaar (Medaillewinnaars)",
        yaxis_title="Percentage met Toekomstige Senioren Prestatie (%)",
        barmode='group',
        yaxis=dict(range=[0, 105]), 
        margin=dict(t=120), 
        font=dict(family="HK Grotesk, Open Sans, sans-serif", size=11),
        height=750,
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="right",
            x=1
        )
    )

    fig.show()
    
    # --- CRITICAL FIX: Return both values so they unpack cleanly ---
    return fig, res_df
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

def plot_cohort_to_medal_conversion(
    df, 
    event_a_list=None, 
    event_b_list=None, 
    start_year=2000,
    end_year=2026,
    gender='All', 
    european_only=False,
    dutch_only=False,
    export_html_path=None
):
    """
    Creates a grouped bar chart showing COHORT CONVERSION TO MEDALS.
    X-Axis: Specific 'A' Tournaments by Year (e.g., Cadet Worlds).
    Y-Axis: The percentage of Top 3, Top 5, Top 7, and Total Participants from 
            Event 'A' who later won a MEDAL (Top 3) in Event 'B' (e.g., Senior Worlds/Olympics).
    Volledig vertaald naar het Nederlands en NOC*NSF huisstijl.
    """
    
    if event_a_list is None:
        event_a_list = ['World Cadet Championships']
    if event_b_list is None:
        event_b_list = ['World Senior Championships', 'Olympic Games']

    df_clean = df.copy()
    df_clean['Year_num'] = pd.to_numeric(df_clean['Year'], errors='coerce')

    # 1. Standaardiseer rangschikking naar niveaus (tiers)
    def get_rank_level(rank_val):
        if pd.isna(rank_val): return 999
        r = str(rank_val).strip().lower()
        if r in ['nan', 'none', '']: return 999
        if r in ['1', '1.0', '1st', '2', '2.0', '2nd', '3', '3.0', '3rd']: return 3
        if r in ['5', '5.0', '5th']: return 5
        if r in ['7', '7.0', '7th']: return 7
        return 99 # Elke andere rank betekent algemene deelname

    df_clean['Rank_Level'] = df_clean['Rank'].apply(get_rank_level)

    # 2. Demografische Filters
    title_filters = []
    if gender and gender.lower() != 'all':
        df_clean = df_clean[df_clean['Gender'].astype(str).str.lower() == gender.lower()]
        nl_gender = "Mannen" if gender.lower() in ['m', 'men', 'male'] else ("Vrouwen" if gender.lower() in ['f', 'w', 'women', 'female'] else gender)
        title_filters.append(f"Geslacht: {nl_gender}")

    if dutch_only:
        dutch_mask = (
            df_clean['Country'].astype(str).str.lower().isin(['netherlands', 'the netherlands', 'ned']) |
            (df_clean['Country Code'].astype(str).str.upper() == 'NED') |
            df_clean['Nationality'].astype(str).str.upper().isin(['NED', 'NETHERLANDS', 'DUTCH'])
        )
        df_clean = df_clean[df_clean['Person'].isin(df_clean[dutch_mask]['Person'].unique())]
        title_filters.append("Nederlandse Atleten")
    elif european_only:
        eju_codes = ['ALB', 'AND', 'ARM', 'AUT', 'AZE', 'BLR', 'BEL', 'BIH', 'BUL', 'CRO', 'CYP', 'CZE', 
                     'DEN', 'EST', 'FIN', 'FRA', 'GEO', 'GER', 'GBR', 'GRE', 'HUN', 'ISL', 'IRL', 'ISR', 
                     'ITA', 'KOS', 'LAT', 'LIE', 'LTU', 'LUX', 'MLT', 'MDA', 'MON', 'MNE', 'NED', 
                     'MKD', 'NOR', 'POL', 'POR', 'ROU', 'RUS', 'SMR', 'SRB', 'SVK', 'SLO', 'ESP', 
                     'SWE', 'SUI', 'TUR', 'UKR']
        euro_mask = (df_clean['Continent'].astype(str).str.lower() == 'europe') | \
                    (df_clean['Country Code'].astype(str).str.upper().isin(eju_codes)) | \
                    (df_clean['Nationality'].astype(str).str.upper().isin(eju_codes))
        df_clean = df_clean[df_clean['Person'].isin(df_clean[euro_mask]['Person'].unique())]
        title_filters.append("Europese Atleten")

    # 3. Filter voor 'A' Evenementen binnen het tijdsbestek
    event_a_df = df_clean[
        (df_clean['Comp.SetDetail'].isin(event_a_list)) & 
        (df_clean['Year_num'] >= start_year) &
        (df_clean['Year_num'] <= end_year)
    ].copy()

    event_a_df['Event_Label'] = event_a_df['Year_num'].astype(int).astype(str) + " " + event_a_df['Comp.SetDetail']
    unique_events = event_a_df[['Year_num', 'Comp.SetDetail', 'Event_Label']].drop_duplicates().sort_values(by=['Year_num', 'Comp.SetDetail'])

    # 4. Extraheer Doel 'B' Historie
    event_b_df = df_clean[df_clean['Comp.SetDetail'].isin(event_b_list)].copy()

    # 5. Verwerk Berekeningen
    results = []
    
    def format_hover_list(athlete_list, chunk_size=3):
        if not athlete_list: return " Geen"
        chunks = [", ".join(athlete_list[i:i+chunk_size]) for i in range(0, len(athlete_list), chunk_size)]
        return "<br>  " + "<br>  ".join(chunks)

    def analyze_cohort(cohort_persons, target_year):
        """Helper to calculate how many of a cohort medalled in B events >= target_year"""
        if len(cohort_persons) == 0: return 0, 0, 0.0, format_hover_list([])
        
        b_results = event_b_df[
            (event_b_df['Person'].isin(cohort_persons)) & 
            (event_b_df['Year_num'] >= target_year)
        ]
        
        if b_results.empty: return len(cohort_persons), 0, 0.0, format_hover_list([])
        
        best_b = b_results.groupby('Person')['Rank_Level'].min()
        medallists = best_b[best_b <= 3].index.tolist()
        
        medal_details = []
        for m in medallists:
            m_yr = b_results[(b_results['Person'] == m) & (b_results['Rank_Level'] <= 3)]['Year_num'].min()
            medal_details.append(f"{m} ({int(m_yr)})")
            
        conv_pct = (len(medallists) / len(cohort_persons)) * 100
        return len(cohort_persons), len(medallists), conv_pct, format_hover_list(medal_details)

    for _, row in unique_events.iterrows():
        c_year = row['Year_num']
        c_event = row['Comp.SetDetail']
        
        a_competitors = event_a_df[(event_a_df['Year_num'] == c_year) & (event_a_df['Comp.SetDetail'] == c_event)]
        
        # Definieer de VIER cohorten
        cohort_t3 = a_competitors[a_competitors['Rank_Level'] <= 3]['Person'].unique()
        cohort_t5 = a_competitors[a_competitors['Rank_Level'] <= 5]['Person'].unique()
        cohort_t7 = a_competitors[a_competitors['Rank_Level'] <= 7]['Person'].unique()
        cohort_part = a_competitors['Person'].unique()

        # Analyseer conversie per cohort naar Evenement B Medailles
        t3_tot, t3_conv, t3_pct, t3_hov = analyze_cohort(cohort_t3, c_year)
        t5_tot, t5_conv, t5_pct, t5_hov = analyze_cohort(cohort_t5, c_year)
        t7_tot, t7_conv, t7_pct, t7_hov = analyze_cohort(cohort_t7, c_year)
        part_tot, part_conv, part_pct, part_hov = analyze_cohort(cohort_part, c_year)

        # Vertaal evenement labels voor de weergave
        nl_event_label = row['Event_Label'].replace("World Cadet Championships", "WK Cadetten").replace("European Judo Championships Cadets", "EK Cadetten").replace("World Junior Championships", "WK Junioren").replace("European Judo Championships Juniors", "EK Junioren")

        results.append({
            'Event_Label': nl_event_label,
            'T3_Total': t3_tot, 'T3_Conv': t3_conv, 'T3_Pct': t3_pct, 'T3_Hov': t3_hov,
            'T5_Total': t5_tot, 'T5_Conv': t5_conv, 'T5_Pct': t5_pct, 'T5_Hov': t5_hov,
            'T7_Total': t7_tot, 'T7_Conv': t7_conv, 'T7_Pct': t7_pct, 'T7_Hov': t7_hov,
            'Part_Total': part_tot, 'Part_Conv': part_conv, 'Part_Pct': part_pct, 'Part_Hov': part_hov
        })

    res_df = pd.DataFrame(results)

    if res_df.empty:
        print("Geen data beschikbaar voor deze specifieke filters.")
        return None, res_df

    # 6. Render Groepering Bar Chart
    fig = go.Figure()

    def build_trace(name, color_rgba, prefix):
        customdata = np.column_stack((
            res_df[f'{prefix}_Total'], 
            res_df[f'{prefix}_Conv'], 
            res_df[f'{prefix}_Pct'], 
            res_df[f'{prefix}_Hov']
        ))
        
        # Formatteer tekst op de balken: bijv. "15%" (alleen tonen als > 0 om visuele ruis te voorkomen)
        text_labels = [f"<b>{pct:.0f}%</b>" if pct > 0 else "" for pct in res_df[f'{prefix}_Pct']]
        
        fig.add_trace(go.Bar(
            name=name,
            x=res_df['Event_Label'],
            y=res_df[f'{prefix}_Pct'],
            marker=dict(color=color_rgba, line=dict(width=0)),
            text=text_labels,
            textposition='inside',
            insidetextanchor='end',
            textfont=dict(size=12, color='white', family="HK Grotesk, Open Sans, sans-serif"),
            customdata=customdata,
            hovertemplate=(
                "<b>%{x}</b><br>"
                f"Basis Cohort (Toernooi A): {name}<br>"
                "Conversie naar Medaille in Doeltoernooi: %{customdata[1]} / %{customdata[0]} (%{customdata[2]:.1f}%)<br><br>"
                "<b>Atleten (Vroegste Medaille Jaar):</b>%{customdata[3]}"
                "<extra></extra>"
            )
        ))

    # Basiskleuren (NOC*NSF Blauw met een 4-stappen transparantieschaal)
    build_trace('Alle Deelnemers', 'rgba(0, 55, 138, 0.25)', 'Part')
    build_trace('Top 7 Eindklassering', 'rgba(0, 55, 138, 0.50)', 'T7')
    build_trace('Top 5 Eindklassering', 'rgba(0, 55, 138, 0.75)', 'T5')
    build_trace('Top 3 (Medaille)', 'rgba(0, 55, 138, 1.0)', 'T3')

    # Titels samenstellen en vertalen
    str_a = ", ".join(event_a_list).replace("World Cadet Championships", "WK Cadetten").replace("European Judo Championships Cadets", "EK Cadetten").replace("World Junior Championships", "WK Junioren").replace("European Judo Championships Juniors", "EK Junioren")
    str_b = ", ".join(event_b_list).replace("World Senior Championships", "WK Senioren").replace("Olympic Games", "Olympische Spelen")
    filter_string = f"[{', '.join(title_filters)}]" if title_filters else "[Allemaal]"

    fig.update_layout(
        title=dict(
            text=(
                f"<b>Conversie naar medaille: {str_a} naar {str_b} Medailles</b><br>"
                f"<span style='font-size:12px; color:gray'>Filters: {filter_string} | Jaren: {start_year}-{end_year}</span>"
            ),
            font=dict(family="HK Grotesk, Open Sans, sans-serif")
        ),
        xaxis_title="Toernooi & Jaar",
        yaxis_title="Percentage met Medaille in Doeltoernooi (%)",
        barmode='group',
        yaxis=dict(range=[0, 105]), 
        margin=dict(t=100), 
        font=dict(family="HK Grotesk, Open Sans, sans-serif", size=11),
        height=750,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
    )

    # Exporteer naar HTML indien het pad is opgegeven
    if export_html_path:
        fig.write_html(export_html_path)
        print(f"Grafiek succesvol opgeslagen naar {export_html_path}")

    fig.show()
    return fig, res_df
```
